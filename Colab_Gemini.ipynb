## Feature Engineering - Categories

### Subtask:
Parse the 'categories' column to extract individual categories. This involves converting the string representation of a list of dictionaries into a proper list of category names and then exploding the DataFrame so each row represents a unique job-category combination.


**Reasoning**:
The first step is to import the `ast` module and define a function that will safely parse the string representation of the list of dictionaries in the 'categories' column and extract the 'category' names. This function will then be applied to the 'categories' column to convert the string data into a list of category names.



```python
import ast

def parse_categories(categories_str):
    if pd.isna(categories_str):
        return []
    try:
        # Safely evaluate the string to a list of dictionaries
        categories_list = ast.literal_eval(categories_str)
        # Extract 'category' names
        return [cat['category'] for cat in categories_list if 'category' in cat]
    except (ValueError, SyntaxError) as e:
        print(f"Error parsing category string: {categories_str}. Error: {e}")
        return []

# Apply the function to the 'categories' column
df['categories'] = df['categories'].apply(parse_categories)

print("Categories column after parsing:")
display(df[['categories']].head())
```

**Reasoning**:
Now that the 'categories' column has been parsed into lists of category names, the next step is to use the `.explode()` method to flatten the DataFrame, creating a new row for each category within a job posting. This will result in each row representing a unique job-category combination as specified in the subtask.



```python
df = df.explode('categories')

print("DataFrame after exploding 'categories' column:")
display(df[['categories']].head())
```

## Feature Engineering - Date Columns

### Subtask:
Convert 'metadata_expiryDate', 'metadata_newPostingDate', and 'metadata_originalPostingDate' columns to datetime objects for easier manipulation and analysis. Additionally, calculate the 'job_age' from the 'metadata_originalPostingDate'.


**Reasoning**:
The first step is to convert the specified date columns to datetime objects, calculate the 'job_age' based on the 'metadata_originalPostingDate' and today's date, and then display the head of the DataFrame and check the data types to verify the transformation.



```python
date_columns = ['metadata_expiryDate', 'metadata_newPostingDate', 'metadata_originalPostingDate']

for col in date_columns:
    df[col] = pd.to_datetime(df[col])

# Calculate 'job_age'
current_date = pd.to_datetime('today').normalize()
df['job_age'] = (current_date - df['metadata_originalPostingDate']).dt.days

print("DataFrame head with new 'job_age' column:")
display(df[['metadata_expiryDate', 'metadata_newPostingDate', 'metadata_originalPostingDate', 'job_age']].head())

print("Data types after conversion:")
df.info()
```

## Analyze Salary Distribution

### Subtask:
Examine the distribution of the 'average_salary' to understand typical salary ranges and analyze average salaries across different 'employmentTypes' and top job categories. Visualize the results using a histogram and bar charts.


**Reasoning**:
I will import the necessary plotting libraries, create a histogram for the 'average_salary' distribution, and then generate bar charts to visualize average salaries across different 'employmentTypes' and the top 10 job categories, following the provided instructions.



```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot style for better aesthetics
sns.set_style("whitegrid")

# 1. Histogram of 'average_salary'
plt.figure(figsize=(10, 6))
sns.histplot(df['average_salary'], bins=50, kde=True)
plt.title('Distribution of Average Salary')
plt.xlabel('Average Salary')
plt.ylabel('Frequency / Count')
plt.show()

# 2. Average Salary by Employment Type
average_salary_by_employment_type = df.groupby('employmentTypes')['average_salary'].mean().sort_values(ascending=False)

plt.figure(figsize=(12, 7))
sns.barplot(x=average_salary_by_employment_type.index, y=average_salary_by_employment_type.values, palette='viridis')
plt.title('Average Salary by Employment Type')
plt.xlabel('Employment Type')
plt.ylabel('Average Salary')
plt.xticks(rotation=45, ha='right') # Rotate labels for readability
plt.tight_layout()
plt.show()

# 3. Average Salary by Top 10 Job Categories
# Ensure top_job_categories is defined from previous steps (it is in kernel state)
top_10_categories_list = top_job_categories.index.tolist()
df_top_categories = df[df['categories'].isin(top_10_categories_list)]
average_salary_by_category = df_top_categories.groupby('categories')['average_salary'].mean().sort_values(ascending=False)

plt.figure(figsize=(12, 7))
sns.barplot(x=average_salary_by_category.index, y=average_salary_by_category.values, palette='plasma')
plt.title('Average Salary by Top 10 Job Categories')
plt.xlabel('Job Category')
plt.ylabel('Average Salary')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

**Reasoning**:
The previous code generated `FutureWarning` messages because `palette` was used without assigning a `hue` argument in `sns.barplot`. To fix this, I will explicitly set `hue` to the x-axis variable and add `legend=False` in both `sns.barplot` calls, which is the recommended practice to ensure the palette is applied correctly without triggering the warning.



```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot style for better aesthetics
sns.set_style("whitegrid")

# 1. Histogram of 'average_salary'
plt.figure(figsize=(10, 6))
sns.histplot(df['average_salary'], bins=50, kde=True)
plt.title('Distribution of Average Salary')
plt.xlabel('Average Salary')
plt.ylabel('Frequency / Count')
plt.show()

# 2. Average Salary by Employment Type
average_salary_by_employment_type = df.groupby('employmentTypes')['average_salary'].mean().sort_values(ascending=False)

plt.figure(figsize=(12, 7))
sns.barplot(x=average_salary_by_employment_type.index, y=average_salary_by_employment_type.values, hue=average_salary_by_employment_type.index, palette='viridis', legend=False)
plt.title('Average Salary by Employment Type')
plt.xlabel('Employment Type')
plt.ylabel('Average Salary')
plt.xticks(rotation=45, ha='right') # Rotate labels for readability
plt.tight_layout()
plt.show()

# 3. Average Salary by Top 10 Job Categories
# Ensure top_job_categories is defined from previous steps (it is in kernel state)
top_10_categories_list = top_job_categories.index.tolist()
df_top_categories = df[df['categories'].isin(top_10_categories_list)]
average_salary_by_category = df_top_categories.groupby('categories')['average_salary'].mean().sort_values(ascending=False)

plt.figure(figsize=(12, 7))
sns.barplot(x=average_salary_by_category.index, y=average_salary_by_category.values, hue=average_salary_by_category.index, palette='plasma', legend=False)
plt.title('Average Salary by Top 10 Job Categories')
plt.xlabel('Job Category')
plt.ylabel('Average Salary')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
